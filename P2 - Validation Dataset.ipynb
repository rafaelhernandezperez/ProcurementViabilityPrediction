```python
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import balanced_accuracy_score
import joblib
```

Now that we have chosen our best predictive model, we have downloaded it and imported it into this notebook to make the predictions with this new data.

```python
loaded_RANDOMFORESTmodel = joblib.load('P2_RANDOMFORESTmodel.joblib')
```

As the assignment stated, we did predictions for low, medium and high competitiveness:

## Low competitiveness

First we imported the dataframe, then dropped the columns we explained on the other notebook that weren't so relevant to this decision process and we filled in the missing values. For the low competitiveness we put 2, because that was our most repeated value in that category in our training data, and the same thing goes for all the other missing values and competitiveness categories to do this comparison.

```python
P_SL = pd.read_csv("MGdataset_P2_validation.csv", decimal=",", delimiter=';')
P_SL = P_SL.drop(columns=['O_daval', 'O_year_p', 'O_countryp_code'])
P_SL["R_bids"]= P_SL ["R_bids"].fillna(2.)
P_SL.to_numpy()
P_SLPred=loaded_RANDOMFORESTmodel.predict(P_SL)
print (P_SLPred)
```

## Medium competitiveness

```python
P_SM = pd.read_csv("MGdataset_P2_validation.csv", decimal=",", delimiter=';')
P_SM = P_SM.drop(columns=['O_daval', 'O_year_p', 'O_countryp_code'])
P_SM["R_bids"]= P_SM ["R_bids"].fillna(3.)
P_SM.to_numpy()
P_SMPred=loaded_RANDOMFORESTmodel.predict(P_SM)
print (P_SMPred)
```

## High Competitiveness

```python
P_SH = pd.read_csv("MGdataset_P2_validation.csv", decimal=",", delimiter=';')
P_SH = P_SH.drop(columns=['O_daval', 'O_year_p', 'O_countryp_code'])
P_SH["R_bids"]= P_SH ["R_bids"].fillna(6.)
P_SH.to_numpy()
P_SHPred=loaded_RANDOMFORESTmodel.predict(P_SH)
print (P_SHPred)
```

# Display our Predictions

Finally this is our result for the predictions in each category:

```python
code = list (range(600,620))

my_predictions = pd.DataFrame({"code": code, "P_SL": P_SLPred, "P_SM": P_SMPred, "P_SH": P_SHPred})

display(my_predictions)
```
